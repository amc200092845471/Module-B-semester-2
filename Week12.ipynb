```python
import matplotlib.pyplot as plt
import pandas as pd

# Simulated data: Before vs After performance for different groups
data = {
    'Group': ['A', 'B', 'C', 'D', 'E'],
    'Before': [70, 55, 60, 40, 50],
    'After': [85, 60, 63, 45, 65]
}

df = pd.DataFrame(data)

# Create the slope graph
fig, ax = plt.subplots(figsize=(8, 6))

# Plot gray lines for all groups
for i in range(len(df)):
    ax.plot(['Before', 'After'], 
            [df.loc[i, 'Before'], df.loc[i, 'After']],
            marker='o', color='gray', linewidth=1.5)
    
    # Left-side label
    ax.text('Before', df.loc[i, 'Before'], 
            f"{df.loc[i, 'Group']} {df.loc[i, 'Before']}", 
            va='center', ha='right', fontsize=9)
    
    # Right-side label
    ax.text('After', df.loc[i, 'After'], 
            f"{df.loc[i, 'After']} {df.loc[i, 'Group']}", 
            va='center', ha='left', fontsize=9)

# Highlight one group (e.g., Group A)
highlight_idx = df[df['Group'] == 'A'].index[0]
ax.plot(['Before', 'After'], 
        [df.loc[highlight_idx, 'Before'], df.loc[highlight_idx, 'After']],
        marker='o', color='#1f77b4', linewidth=2.5)

# Bold and color labels for highlighted group
ax.text('Before', df.loc[highlight_idx, 'Before'], 
        f"A {df.loc[highlight_idx, 'Before']}", 
        va='center', ha='right', fontsize=10, fontweight='bold', color='#1f77b4')
ax.text('After', df.loc[highlight_idx, 'After'], 
        f"{df.loc[highlight_idx, 'After']} A", 
        va='center', ha='left', fontsize=10, fontweight='bold', color='#1f77b4')

# Remove clutter and format axes
ax.set_title("Performance Before and After", fontsize=14, weight='bold')
ax.set_xticks(['Before', 'After'])
ax.set_xlim(-0.5, 1.5)
ax.set_ylim(30, 90)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.tick_params(left=False, bottom=False)
ax.set_yticks([])

plt.tight_layout()
plt.show()
```
