# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
import numpy as np

# Synthetic example data – change in percentage over time
categories = ['Product A', 'Product B', 'Product C', 'Product D', 'Product E']
start_values = [35, 55, 42, 60, 25]  # Values for year 2023
end_values   = [50, 40, 60, 58, 20]  # Values for year 2024

fig, ax = plt.subplots(figsize=(10, 6))

# Draw lines and labels for each category
for i, (start, end) in enumerate(zip(start_values, end_values)):
    color = 'firebrick' if end > start else 'dimgray'  # Highlight increase
    ax.plot([0, 1], [start, end], marker='o', color=color, linewidth=2.5)
    
    # Left label (year 2023)
    ax.text(-0.05, start, f"{categories[i]} {start}%", ha='right', va='center',
            fontsize=10, color=color)
    
    # Right label (year 2024)
    ax.text(1.05, end, f"{end}%", ha='left', va='center',
            fontsize=10, color=color)

# Add year labels at the top
ax.text(0, max(start_values) + 10, "2023", ha='center', fontsize=12, weight='bold', color='dimgray')
ax.text(1, max(end_values) + 10, "2024", ha='center', fontsize=12, weight='bold', color='dimgray')

# Chart formatting
ax.set_xlim(-0.5, 1.5)
ax.set_ylim(0, max(start_values + end_values) + 20)
ax.axis('off')  # Remove axes and ticks
ax.set_title("Change in Product Adoption from 2023 to 2024",
             fontsize=16, weight='bold', loc='left')

plt.tight_layout()
plt.show()
```
